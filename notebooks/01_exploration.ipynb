```python
print("Lignes identiques :", df.duplicated().sum())
cle = "id_pdc_itinerance"
print("Clé vide :", df[cle].isna().sum())
print("Clés répétées :", df[cle].duplicated(keep=False).sum())
print("Clés distinctes :", df[cle].nunique())
```

```python
for col in ["paiement_cb", "gratuit", "cable_t2_attache"]:
    print(col, df[col].value_counts(dropna=False).to_dict())
print(df["num_pdl"].value_counts().head())
print(df["restriction_gabarit"].value_counts().head())
print(df["code_insee_commune"].str.len().value_counts().head())
```

```python
d = df[df["id_pdc_itinerance"].duplicated(keep=False)]
diff = d.groupby("id_pdc_itinerance").nunique(dropna=False)
print((diff > 1).sum().sort_values(ascending=False).head(10))
```

```python
import pandas as pd

tri = df.assign(_d=pd.to_datetime(df["date_maj"], errors="coerce"))
tri = tri.sort_values("_d", ascending=False, na_position="last")
sans_doublons = tri.drop_duplicates("id_pdc_itinerance", keep="first").drop(
    columns="_d"
)
print(len(sans_doublons))
```
